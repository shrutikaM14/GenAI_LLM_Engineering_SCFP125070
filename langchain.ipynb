Pydantic output parser

```python
from langchain_core.output_parsers import PydanticOutputParser
```

```python
from pydantic import BaseModel, Field
```

```python
class Student(BaseModel):
    name: str = Field(description="Student name")
    roll_no: int = Field(description="Student roll no")
    batch: str = Field(description="Student acedemic batch")
    cgpa: float = Field(description="Student CGPA")
```

```python
pydantic_parser = PydanticOutputParser(pydantic_object=Student)
```

```python
print(pydantic_parser.get_format_instructions())
```

```python
input_temp = ChatPromptTemplate([
    ("system","You are a helpful ai assistant.extract student information into thisw formation {PYDANTIC_FORMATTING}"),
    ("user","Student information is {student_info}")   
]).partial(PYDANTIC_FORMATTING = pydantic_parser.get_format_instructions())
```

```python
input_temp
```

```python
input = input_temp.invoke({"student_info":"student name is vaishnavi and student roll num is 63, batch is 2025 cgpa is 8.5"})
```

```python
response = llm.invoke(input)
```

```python
response
```

```python
json_parser.invoke(response)
```
