```python
import torch

print("PyTorch version:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
    print("GPU memory:", round(torch.cuda.get_device_properties(0).total_memory / 1024**3, 2), "GB")
else:
    print("GPU is not being used.")
```

```python
import torch

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

x = torch.randn(5000, 5000, device=device)
y = torch.matmul(x, x)

print("Device used:", x.device)
print("GPU memory allocated:", round(torch.cuda.memory_allocated() / 1024**3, 2), "GB")
print("GPU memory reserved:", round(torch.cuda.memory_reserved() / 1024**3, 2), "GB")
```

```python
import torch

for size in [1000, 2000, 4000, 6000]:
    torch.cuda.empty_cache()

    x = torch.randn(size, size, device="cuda")
    y = torch.matmul(x, x)

    memory = torch.cuda.memory_allocated() / 1024**3

    print(f"Matrix size: {size}x{size}")
    print(f"GPU memory allocated: {memory:.2f} GB")
    print("-" * 30)

    del x, y
```
